# 旺店通 SDK 测试

凭证从同目录的 `wangdian_config.py` 读取，该文件已加入 `.gitignore`。

```python
import sys
from pathlib import Path

cwd = Path.cwd().resolve()
project_root = cwd if (cwd / 'pyproject.toml').exists() else cwd.parent
examples_dir = project_root / 'examples'
sys.path.insert(0, str(project_root / 'src'))
sys.path.insert(0, str(examples_dir))

from wangdian import ApiError, WangdianClient
from wangdian_config import APP_KEY, APP_SECRET, ENVIRONMENT, SID

if not all((SID, APP_KEY, APP_SECRET)):
    raise ValueError('请先填写 examples/wangdian_config.py 中的 SID、APP_KEY 和 APP_SECRET')

client = WangdianClient(
    sid=SID,
    app_key=APP_KEY,
    app_secret=APP_SECRET,
    environment=ENVIRONMENT,
    timeout=30,
)
print(f'当前环境: {client.environment.value}')
print(f'请求网关: {client.base_url}')
```

## 查看签名后的参数（不发送请求）

```python
params = client.build_parameters({
    'page_no': 0,
    'page_size': 100,
})
{**params, 'appkey': '***', 'sign': '***'}
```

## 查询店铺

先用读取类接口验证账号和签名。测试环境需要使用旺店通分配的测试凭证。

```python
try:
    shops = client.call('shop', {'page_no': 0, 'page_size': 100})
    shops
except ApiError as exc:
    print(f'旺店通错误 code={exc.code}: {exc.message}')
    print(exc.response)
print(shops)
```

## 查询订单示例

```python
trade_result = client.call(
    'trade_query',
    {
        'start_time': '2026-07-29 00:00:00',
        'end_time': '2026-07-29 23:59:59',
        'page_no': 0,
        'page_size': 40,
    },
)
trade_result
```
